```python
def clean_distances(miles_list):
  """ Return a new list containing only non-negative measurements.
  Do not modify the original list. """
  distances = []
  for distance in miles_list:
    if distance >= 0:
      distances.append(distance)
  return distances
```

```python
miles_list = [5, -3, 10, 0, -2, 8]
distances = clean_distances(miles_list)
print(distances)
```

```python
# Test 1: Mixed values
miles_list = [5, -3, 10, 0, -2, 8]
assert clean_distances(miles_list) == [5, 10, 0, 8]

# Test 2: Empty list
miles_list = []
assert clean_distances(miles_list) == []

# Test 3: All negative values
miles_list = [-1, -5, -10]
assert clean_distances(miles_list) == []

# Test 4: Original list unchanged

original = [3, -2, 7]
backup = original.copy()

result = clean_distances(original)

assert original == backup
assert result is not original

print ("ALL tests passed!")
```

Day-Data Cleaning

Project Description

 This Python program cleans a list of distances by removing negative numbers while keeping zero and positive numbers.

 How My Code works

 1.The fuction clean_distances() receives a list of distances.
 2. It creates an empty list to store valid distances.
 3. A for loop checks each number.
 4. An if statement determines whether the number is greater than or equal to zero.
 5. The .append() method adds valid numbers to the new list.
 6. The function returns the cleaned list without modifying the original data.
