```python
def convert_to_km(distances):
    """ Convert a list of distances from miles to kilometers."""
    distance_km = []
    for i in range(len(distances)):
        km = distances[i] * 1.60934
        distance_km.append(km)
    return distance_km
```

```python
assert convert_to_km([1]) == [1.60934]
assert convert_to_km([]) ==[]

print ("Both tests passed!")
```

```python
original = [1, 5, 10]
backup = original.copy()

result = convert_to_km(original)

assert original == backup

print ("Original list preserved!")
```
