```python
def clean_distances(miles_list):
  """ Return a new list containing only non-negative measurements.
  Do not modify the original list. """
  distances = []
  for distance in miles_list:
    if distance >= 0:
      distances.append(distance)
  return distances
```

```python
import csv

def export_clean_distances(miles_list, filename):
  """ Clean distance measurements and export valid values to a csv file.

  Preserve the original input list."""
  distances = clean_distances(miles_list)
  with open(filename, "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(["distance_miles"])
    for d in distances:
      writer.writerow([d])
  return distances
```

### The  fuction Clean_distances:

- clean a list of distances
- create a CSV file
- write each distances in separate row
- return the clean list of distances

```python
# Test 1
original = [5, -3, 10, 0, -2, 8]
backup = original.copy()

export_clean_distances(original, "distances.csv")

with open("distances.csv", "r", encoding="utf-8") as file:
  rows = list(csv.reader(file))
  assert rows == [
      ["distance_miles"],
      ["5"],
      ["10"],
      ["0"],
      ["8"]
  ]

  assert original == backup
  print("CSV export test passed!")
```

```python
# Test
empty = []
export_clean_distances(empty, "empty.csv")

with open("empty.csv", "r", newline="", encoding="utf-8") as file:
  rows = list(csv.reader(file))
  assert rows == [["distance_miles"]]
  print("Empty list test passed!")
```
